```python
import numpy as np
from sklearn.linear_model import LinearRegression

hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
scores = np.array([38, 45, 55, 60, 72, 75, 88, 92])

# 1단계: 라이브러리로 한 번에
model = LinearRegression()
model.fit(hours.reshape(-1, 1), scores)
print("기울기 a:", model.coef_[0], "절편 b:", model.intercept_)
print("5.5시간 공부하면:", model.predict([[5.5]])[0])

# 2단계: 경사 하강법으로 직접
a, b = 0.0, 0.0
lr = 0.01  # 한 번에 움직이는 보폭(학습률)
for epoch in range(5001):
    pred = a * hours + b
    error = pred - scores
    loss = np.mean(error ** 2)          # 오차 계산 (MSE)
    a -= lr * np.mean(2 * error * hours)  # 매개변수 조금씩 수정
    b -= lr * np.mean(2 * error)
    if epoch in (0, 10, 100, 1000, 5000):
        print(f"epoch {epoch:>4} | loss {loss:>8.2f} | a {a:.3f} b {b:.3f}")
```
